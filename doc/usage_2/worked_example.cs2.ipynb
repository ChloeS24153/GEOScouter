## **Worked Example 2: End to End GEOscouter Run**
---
This guide provides a reproducible end-to-end workflow to test GEOscouter using a fixed input and expected outputs.

#### **Pinned Parameters**

| Parameter | Value |
|-----------|-------|
| **Input File** | [`kidney_cosmx.txt`](kidney_cosmx.txt) |
| **Step 2 - Platform** | cosmx |
| **Step 3 - Signature Tokens** | exprMat_file.csv.gz, metadata_file.csv.gz, fov_positions_file.csv.gz, polygons.csv.gz, tx_file.csv.gz |
| **Step 4 - Similarity Threshold** | 1 |
| **Step 5 - Keyword Search** | cosmx |

### **Step 1 - Run data scraping**
1. Upload the file, [`kidney_cosmx.txt`](kidney_cosmx.txt), making sure to delete cached outputs from previous runs
2. Click Run Pipeline
3. The app will scrape GEO for the 12 GSEs found in the input. Once complete, you can download the raw scrape results (e.g., geo_webscrap.csv).

### **Step 2 - Filter datasets**
Filters can be used to narrow down the GSEs, using "Assay type" and "Platform (GPL)". This ensures you only get the GSEs you want to work with, and not other platforms. The minimum and maximum number of samples can also be filtered.

To filter down in the assay type, select all that mention cosmx. Leave the minimum and maximum samples then press "Apply scrape level filters". 

### **Step 3 - Visualise datasets**
This sections allow you to visualise each GSE based on three categories, the file size, number of samples and the number of file types. The barplots allow you to compare the differences between datasets, to identify high quality ones. You can click **"Generate all visualisations**"** to produce dataset snapshots. 

1. Displays the total size of each GSE



2. Displays the number of samples for each GSE



3. Displays the number of file types for each GSE



#### **File vs Sample Complexity**



All visualisations can be downloaded as a PNG, by clicking on the camera button on top of the figure wanted. 

### **Similarity Network (supervised file-structure signature network)**

1. Select Manual file types as the signature source.
2. Enter the signature tokens listed in the parameters table (one per line).
3. Click Build signature from file types to review the match patterns with their weights. 
4. Click on apply signature and show network to display all filtered GSE datasets in comparison.
5. When reviewing the signature patterns, the weight of each file type can be changed before the network is produced. 

However, a signature can be produced by a **training GSE**, where you can enter a GSE from the dataset with the required and wanted file types. That GSE will then be used to compare the rest of the datasets to identify similarity.



The threshold of the minimum similarity of GSEs can be changed with the slider. The network will change and the key to the right shows the signature similarity, however in this example there are 5 samples that have a similarity of 1 and the rest are not similar at all. 


### **Step 4 - Select specific GSEs**
Here we select the GSEs we want to keep for the final comparison. You can add datasets in bulk based on a minimal value of signature similarity, or you can manually enter specific GSE IDs if you want to force their inclusion regardless of their score.

1. Slide the Minimum signature similarity threshold to **1.0**
 and click Add GSEs matching signature (>= threshold). This automatically adds datasets that structurally match your signature rules at 100%.
2. To add a dataset manually (for example, a dataset that scored below the threshold but is relevant to your analysis), enter the GSE ID in the text field and click Add manual GSE. For this example, no manually added GSEs are used.
3. The comparison list now contains **6 unique GSEs**.
4. Click Export comparison list as filtered table to download the subset containing basic information of the filtered GSEs.
Export: **filtered_geo_webscrap.csv**



### **Step 5 - Sample metadata filtering and curation**

Now we perform sample metadata filtering and curation strictly for the selected GSEs.

1. First, we can filter specific information from the metadata (such as organ/tissue, disease, etc.). In this example, we are applying no filtering and moving to the next part.
2. The curated GSE catalog appears, showing the metadata of each GSE after applying filters.
Export: **curated_gse_catalogue.csv**
3. Next, we can view the metadata of each GSM separately. You can also search for keywords to find samples that mention a specific word anywhere in their metadata. The app returns the number of GSEs and GSMs containing that keyword.
4. Using "Cosmx" as an example keyword yields 6 GSEs and 158 GSMs.
5. Click Export all fetched metadata to Excel. This creates an Excel workbook where each tab is one GSE, containing the metadata of all its samples.
Export: **metadata_GSE.xlsx**
